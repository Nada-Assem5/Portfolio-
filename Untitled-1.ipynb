```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv(r"C:\Users\عالم التكنولوجيا\Downloads\archive (4)\StudentsPerformance.csv")
```

```python
data.sample(10)
```

```python
data.info()
```

```python
data.describe()
```

```python
for i in data.columns:
    print(f"\nColumn: {i}")
    print(data[i].value_counts())
```

```python
data['average_score'] = data[['math score', 'reading score', 'writing score']].mean(axis=1)
```

```python
score_cols = ['math score', 'reading score', 'writing score', 'average_score']
long_df = data.melt(id_vars='gender', value_vars=score_cols, 
                    var_name='Score Type', value_name='Score')

plt.figure(figsize=(12,10))
sns.barplot(data=long_df, x='gender', y='Score', hue='Score Type', palette='viridis')
plt.title('Scores by Gender and Score Type')
plt.xlabel('Gender')
plt.ylabel('Score')
plt.legend(title='Score Type')
plt.show()
```

```python
sns.pairplot(data, vars=['math score', 'reading score', 'writing score'], hue='gender', palette='husl')
plt.suptitle('Pairplot of Scores by Gender', y=1.02)
plt.show()
```

```python
import numpy as np

plt.figure(figsize=(8,6))
corr = data[['math score', 'reading score', 'writing score', 'average_score']].corr()

sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap of Scores')
plt.show()
```

```python

plt.figure(figsize=(10,5))
sns.barplot(data=data, x='parental level of education', y='average_score', palette='crest', estimator='mean')
plt.title('Average Student Score by Parental Education Level')
plt.xlabel('Parental Education Level')
plt.ylabel('Average Score')
plt.xticks(rotation=30)
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.barplot(data=data, x='parental level of education', y='average_score', hue='gender', palette='husl', estimator='mean')
plt.title('Average Score by Parental Education Level and Gender')
plt.xlabel('Parental Education Level')
plt.ylabel('Average Score')
plt.xticks(rotation=30)
plt.legend(title='Gender')
plt.show()
```
